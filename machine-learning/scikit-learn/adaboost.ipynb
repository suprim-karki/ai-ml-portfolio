```python
import pandas as pd
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=2000, n_features=10, n_informative=8, n_redundant=2, random_state=11)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=11)
```

```python
from sklearn.ensemble import AdaBoostClassifier

ada = AdaBoostClassifier()
```

```python
ada.fit(X_train, y_train)

y_pred = ada.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

## Ada Boost with Logistic Regression and SVM

```python
from sklearn.linear_model import LogisticRegression

adalog = AdaBoostClassifier(estimator=LogisticRegression())
```

```python
adalog.fit(X_train, y_train)

y_pred2 = adalog.predict(X_test)
```

```python
adalog.fit(X_train, y_train)

y_pred2 = adalog.predict(X_test)
```

```python
print(confusion_matrix(y_test, y_pred2))
print(classification_report(y_test, y_pred2))
```

```python
from sklearn.svm import SVC

svc = SVC(kernel='linear', probability=True)
```

```python
adasvm = AdaBoostClassifier(estimator=svc, n_estimators=25, learning_rate=0.1)

adasvm.fit(X_train, y_train)

y_pred3 = adasvm.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print(confusion_matrix(y_test, y_pred3))
print(classification_report(y_test, y_pred3))
```
