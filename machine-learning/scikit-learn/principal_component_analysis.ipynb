```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('2022mlbteams.csv')

df.head()
```

```python
df.drop('Tm', axis=1, inplace=True)

df.head()
```

```python
X = df.iloc[:,0:27]
y = df.iloc[:,27]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=21)
```
